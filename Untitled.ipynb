```python
from scipy.signal import find_peaks

cases = india_daily["new_cases_smoothed"].fillna(0)

peaks, properties = find_peaks(
    cases,
    prominence=20000,
    distance=60
)

case_peaks = india_daily.iloc[peaks][
    ["date", "new_cases_smoothed"]
].sort_values(
    "new_cases_smoothed",
    ascending=False
)

case_peaks
```

```python
deaths = india_daily["new_deaths_smoothed"].fillna(0)

death_peaks, death_properties = find_peaks(
    deaths,
    prominence=200,
    distance=60
)

death_peak_table = india_daily.iloc[death_peaks][
    ["date", "new_deaths_smoothed"]
].sort_values(
    "new_deaths_smoothed",
    ascending=False
)

death_peak_table
```

```python
case_peak_date = case_peaks.iloc[0]["date"]

death_peak_date = death_peak_table.iloc[0]["date"]

lag_days = (
    death_peak_date - case_peak_date
).days

print("Case peak:", case_peak_date)
print("Death peak:", death_peak_date)
print("Lag:", lag_days, "days")
```

```python
wave_vaccination = []

for _, row in wave_summary.iterrows():

    peak_date = row["case_peak_date"]

    # Find the closest available vaccination record
    vaccine_row = india_vaccine.iloc[
        (india_vaccine["date"] - peak_date).abs().argsort()[:1]
    ]

    wave_vaccination.append({
        "case_peak_date": peak_date,
        "peak_cases_7day_avg": row["peak_cases_7day_avg"],
        "vaccination_date": vaccine_row.iloc[0]["date"],
        "vaccination_coverage": vaccine_row.iloc[0]["vaccination_coverage"],
        "people_fully_vaccinated": vaccine_row.iloc[0]["people_fully_vaccinated"]
    })

wave_vaccination = pd.DataFrame(wave_vaccination)

wave_vaccination
```

```python
major_waves = wave_summary[
    wave_summary["case_peak_date"] >= "2021-01-01"
].copy()

major_waves
```

```python
major_waves_vaccine = major_waves.merge(
    india_vaccine[
        [
            "date",
            "people_fully_vaccinated",
            "vaccination_coverage"
        ]
    ],
    left_on="case_peak_date",
    right_on="date",
    how="left"
)

major_waves_vaccine[
    [
        "case_peak_date",
        "peak_cases_7day_avg",
        "people_fully_vaccinated",
        "vaccination_coverage"
    ]
]
```

```python
may_2021 = major_waves_vaccine.iloc[0]
jan_2022 = major_waves_vaccine.iloc[1]

vaccination_change = (
    jan_2022["vaccination_coverage"]
    - may_2021["vaccination_coverage"]
)

case_change = (
    (jan_2022["peak_cases_7day_avg"]
     - may_2021["peak_cases_7day_avg"])
    / may_2021["peak_cases_7day_avg"]
) * 100

print(f"Increase in vaccination coverage: {vaccination_change:.2f} percentage points")
print(f"Change in peak cases: {case_change:.2f}%")
```
